```python
#13.	Write a function grade(score) that returns:
#a.	90 and above: "Distinction"
#b.	75 to 89: "First Class"
#c.	60 to 74: "Second Class"
#d.	below 60: "Fail"

def grade(score):
  if score>=90:
    return"distinction"
  elif score>=75 and score<=89:
    return"first class"
  elif score>=60 and score<=74:
    return"second class"
  else:
      return"fail"
print("anu:",grade(78))
print("binu:",grade(85))
print("cathy:",grade(92))
print("deepa:",grade(67))
print("ebin:",grade(74))
print("fathima:",grade(88))
print("gopal:",grade(95))
print("hari:",grade(61))







```
